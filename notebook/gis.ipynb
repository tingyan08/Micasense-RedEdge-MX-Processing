```python
import cv2, sys
import numpy as np
import geopandas as gpd
import os
from micasense import metadata
sys.path.append("d:/USDA-NIFA/Code")
os.chdir("d:/USDA-NIFA/Code")
os.getcwd()
```

```python
def transform2seconds(time_str):
    """Convert a time string in the format 'HH:MM:SS' to total seconds."""
    h, m, s = map(float, time_str.split(':'))
    return h * 3600 + m * 60 + s
```

```python
root_folder = "B1"
image_folder = os.path.join(root_folder, "Processed", "capture")
geojason_file = os.path.join(root_folder, "Processed", "imageSet.json")
UTM_CRS = "EPSG:32616" # UTM Zone 16N for Indiana

gdf = gpd.read_file(geojason_file)
gdf['img_path'] = gdf['capture_id'].apply(lambda x: os.path.normpath(os.path.join(image_folder, f"{x}.tif")))
gdf['capture_time'] = gdf['capture_id'].apply(lambda x: transform2seconds(metadata.Metadata(os.path.normpath(os.path.join(image_folder, f"{x}.tif"))).get_item("EXIF:GPSTimeStamp")))

gdf_utm = gdf.to_crs(UTM_CRS)
```

```python
gdf.to_file('B1/B1.geojson', driver='GeoJSON')
```

```python
gdf.sort_values('capture_time', inplace=True)
```
