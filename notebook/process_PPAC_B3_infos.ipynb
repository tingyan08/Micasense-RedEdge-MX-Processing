```python
df = pd.read_excel("Data/090523_PPAC_B3/PPAC.xlsx", sheet_name="GCP Coordinates", header=None)
```

```python
df
```

```python
info = df.iloc[:, 1].to_list()
info = [i for i in info if str(i) != "nan"]
```

```python
gcp = {"name":[], "latitude":[], "longitude":[]}
for i in range(0, len(info), 3):
    gcp["name"].append(f"GCP_{int(info[i])}")
    gcp["latitude"].append(info[i+1])
    gcp["longitude"].append(info[i+2])
```

```python
gcp_df = pd.DataFrame(gcp)
gcp_df.to_csv("Data/PPAC_B3_gcp.csv", index=False, header=False)
```
