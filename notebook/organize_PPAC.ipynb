```python
import pandas as pd
from pyproj import Proj, transform
```

```python
import re

def dms_to_decimal(dms: str) -> float:
    d, m, s = map(float, re.findall(r"[\d.]+", dms))
    sign = -1 if re.search(r"[SW]", dms, re.I) else 1
    return sign * (d + m / 60 + s / 3600)
```

```python
aoi_path = "../Data/PPAC-B3/2021/PPAC_B3_2021_Grid_Coordinates_Center.xlsx"
```

```python
df = pd.read_excel(aoi_path, header=0)
```

```python
df.drop_duplicates(subset=['Grid ID'], inplace=True)
```

```python
df
```

```python
df = df[df["Grid ID"].apply(lambda x: isinstance(x, str) and re.fullmatch(r"[A-I]\d{1,2}", x) is not None)].copy()
df["Latitude (Center)"] = df["Latitude (Center)"].apply(dms_to_decimal)
df["Longitude (Center)"] = -df["Longitude (Center)"].apply(dms_to_decimal).abs()
df
```

```python
def grid_id_to_point_id(gid: str) -> int:
    row = 10 - (ord(gid[0].upper()) - ord("A") + 1)
    col = int(gid[1:])
    return (row - 1) * 15 + col

aoi = pd.DataFrame({
    "Point_ID": df["Grid ID"].apply(grid_id_to_point_id).values,
    "Latitude": df["Latitude (Center)"].values,
    "Longitude": df["Longitude (Center)"].values,
})
aoi
```

```python
aoi.to_csv("../Data/PPAC-B3/2021/PPAC-B3_aoi.csv", index=False)
```

```python
import json
from pyproj import Transformer

transformer = Transformer.from_crs("EPSG:4326", "EPSG:32616", always_xy=True)
half = 6.0

features = []
for _, r in aoi.iterrows():
    e, n = transformer.transform(r["Longitude"], r["Latitude"])
    features.append({
        "type": "Feature",
        "properties": {"id": int(r["Point_ID"]), "center": f"({e}, {n})"},
        "geometry": {
            "type": "Polygon",
            "coordinates": [[
                [e + half, n - half],
                [e + half, n + half],
                [e - half, n + half],
                [e - half, n - half],
                [e + half, n - half],
            ]],
        },
    })

geojson = {
    "type": "FeatureCollection",
    "name": "aoi_square",
    "crs": {"type": "name", "properties": {"name": "urn:ogc:def:crs:EPSG::32616"}},
    "features": features,
}

out_path = "../Data/PPAC-B3/2021/PPAC-B3_aoi_square.geojson"
with open(out_path, "w") as f:
    json.dump(geojson, f)
print(f"Wrote {len(features)} features to {out_path}")
```
