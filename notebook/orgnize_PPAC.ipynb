```python
for folder in micasense_folders:
    tiff_images = glob.glob(os.path.join(folder, '**/*.tif'), recursive=True)
    if 2000 > len(tiff_images) > 5:
        print(f'Found {len(tiff_images)} tiff images in {folder}')
        date = folder.split('/')[5]
        print(f'Extracted date: {date}')
        # YYYYMMDD to MMDDYY
        date = date[4:6] + date[6:8] + date[2:4] + "_PPAC-B3"
        os.makedirs('/media/tingyan/Research/USDA-NIFA/Code/Data/PPAC-B3/' + date, exist_ok=True)
        os.makedirs(f'/media/tingyan/Research/USDA-NIFA/Code/Data/PPAC-B3/{date}/Images', exist_ok=True)
        os.makedirs(f'/media/tingyan/Research/USDA-NIFA/Code/Data/PPAC-B3/{date}/Panel', exist_ok=True)
        for image in tiff_images:
            shutil.copy(image, f'/media/tingyan/Research/USDA-NIFA/Code/Data/PPAC-B3/{date}/Images/')
```
