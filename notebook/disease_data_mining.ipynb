```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
disease_data = pd.read_csv("../Data/PPAC-B3/Disease/2024_PPAC_B3_VisualEvaluations_Mean_Calibrated.csv")
disease_data.drop(columns=["cn12", "cn13", "tarcn12", "tarcn13"], inplace=True)
vi_data = pd.read_excel("../combined_vi_stats_filtered_PPAC-B3.xlsx")
aoi_data = pd.read_csv("../Data/PPAC-B3/PPAC-B3_aoi.csv")
```

#### Filter with canopy

```python
process_disease_data = disease_data[disease_data["leaf"] == "E2"].fillna(0)
process_disease_data
```

#### Add Point_ID to disease_data

```python
# Merge the two DataFrames on their coordinate columns
updated_disease_data = pd.merge(
    process_disease_data,
    aoi_data[["row", "column", "Point_ID"]],
    left_on=["x", "y"],
    right_on=["row", "column"],
    how="left",
)
```

```python
updated_disease_data
```

#### Draw the trends between each indices and disease percentage

```python
disease_date = ["06/18/2024", "06/27/2024", "07/02/2024", "07/08/2024", "07/17/2024", "07/23/2024", "08/07/2024", "08/14/2024", "08/20/2024", "08/30/2024", "09/06/2024", "09/13/2024"]
```

```python
### Modify date format
vi_data["DATE"] = pd.to_datetime(vi_data["DATE"], format="%m%d%Y").dt.strftime("%m/%d/%Y")
```

```python
dates = vi_data["DATE"].unique()
methods = vi_data["METHOD"].unique()
soils = vi_data["SOIL"].unique()
```

```python
vi_data
```

```python
import matplotlib.dates as mdates
import os
os.makedirs("../temp/", exist_ok=True)

# Average disease severity across all 135 points at each of the 12 disease dates
disease_dates_dt = pd.to_datetime(disease_date, format="%m/%d/%Y")
avg_disease = updated_disease_data.iloc[:, 4:16].mean(axis=0).values

# Only MEAN VI columns
vi_cols = [c for c in vi_data.columns[7:] if "_MEAN" in c]

colors = ["#e41a1c", "#377eb8", "#4daf4a", "#984ea3"]

for vi in ["OSAVI_MEAN"]:
    fig, ax1 = plt.subplots(figsize=(12, 5))

    ax1.plot(disease_dates_dt, avg_disease, "k-o", lw=2, ms=5, label="Avg Disease Severity")
    ax1.set_ylabel("Disease Severity")
    ax1.set_xlabel("Date")
    ax1.set_ylim(bottom=0)

    ax2 = ax1.twinx()
    for ci, (method, soil) in enumerate([(m, s) for m in methods for s in soils]):
        grp = vi_data[(vi_data["METHOD"] == method) & (vi_data["SOIL"] == soil)]
        avg_vi = grp.groupby("DATE")[vi].mean().reset_index()
        avg_vi["DATE"] = pd.to_datetime(avg_vi["DATE"], format="%m/%d/%Y")
        avg_vi = avg_vi.sort_values("DATE")
        ax2.plot(avg_vi["DATE"], 1 - avg_vi[vi], "-s", color=colors[ci],
                 lw=1.5, ms=4, label=f"{method}/{soil}")
    ax2.set_ylabel(vi)

    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper left", fontsize=9)

    ax1.xaxis.set_major_formatter(mdates.DateFormatter("%m/%d"))
    fig.autofmt_xdate(rotation=45)
    plt.title(f"{vi} vs Disease Severity (avg across all points)")
    plt.tight_layout()
    plt.savefig(f"../temp/summary_{vi}_vs_disease.png", dpi=150)
    plt.close()

print(f"Saved {len(vi_cols)} plots to ../temp/")
```
